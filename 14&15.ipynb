```python
def countDog(st):
  count = 0
  for word in st.lower().split():
    if word == 'dog':
      count += 1
  return count
```

```python
countDog('this dog runs faster than the othe dog dude!')
```

```python
def caught_speeding(speed,is_birthday):

  if is_birthday:
    speeding = speed - 5
  else:
    speeding = speed

  if speeding >80:
    return ' Big ticket'
  elif speeding > 60:
    return 'Samll ticket'
  else:
    return 'No ticket'
```

```python
caught_speeding(81,True)
```

```python
caught_speeding(81,False)
```
